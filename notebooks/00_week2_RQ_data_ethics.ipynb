Knowing which properties predict a high Tc could help guide the search for new materials.]

## Research Questions and Answers

For every answer, distinguish between **known evidence**, **reasonable interpretation**, and **unknown information to investigate**.

### 1. Where did the data come from?

Known: Compiled by Kam Hamidieh (2018) from the NIMS SuperCon database in Japan. SuperCon is a record of superconductors reported in the published literature. Hamidieh calculated 81 features from the properties of each material’s elements. Hosted on UCI (DOI: 10.24432/C53P47), CC BY 4.0. Accessed 2026-10-11 through the Kaggle mirror (munumbutt/superconductor-dataset).

Interpretation: The Kaggle copy matches the UCI version: same shape (21,263 × 82) and the same statistics.

Unknown: Exactly how SuperCon checks the Tc values it records, and whether the Kaggle uploader changed anything.


### 2. Why was it originally collected?

Known: SuperCon was built to record superconductors from research papers. Hamidieh compiled this version to build a statistical model that predicts Tc (Hamidieh, 2018).

Interpretation: Our use is the same kind of research, predicting and explaining Tc, plus a new classification task (above or below 77 K). The change is small and ethically justified: it is non-commercial academic work, the data is public, and the licence allows reuse with credit.

Unknown: N/A.


### 3. Who is represented?

Indirectly: Hamidieh may be indirectly represented, as his choices in compiling the dataset, such as which SuperCon entries to keep and which features to calculate, shape the data and so our results.

Known: No people. Each row is one superconducting material, described by its composition and measured Tc. About half are copper oxides (cuprates); the rest include alloys, iron-based compounds and others.

Interpretation: The data represents materials that were synthesised, measured and published, mostly from university and industry labs in research-active countries.

Unknown: The exact time span and how the materials are split by country or lab.


### 4. Who might be missing?

Known:
No non-superconductors or failed experiments.
No pressure or crystal-structure information.
Materials found after the dataset was compiled are missing (e.g. nickelates, 2019).
Only about 15,500 unique formulas, with repeats from different papers.

Interpretation: Publication bias. High-Tc and “interesting” materials, especially cuprates, are overrepresented, and negative results are underrepresented. Labs with fewer resources or that publish outside major journals may also be underrepresented.

Unknown: Whether materials were dropped because element data was missing, and how many.


### 5. What personal or sensitive information does it contain, explicitly or implicitly?

N/A. There are no personal identifiers, sensitive fields or human subjects. There are no author or lab names in the files, so no combination of fields could identify a person.

### 6. What could your analysis reveal that was not obvious in the original data?

Known (from our analysis):
99.7% of materials above 77 K are cuprates.
Tc is strongly linked to features like atomic radius range, valence and thermal conductivity.
5,703 rows repeat a formula already in the data.

Interpretation: These are statistical associations, not causes. A model might seem to “predict” high Tc when it is mostly detecting copper.

Unknown: Whether these patterns hold for new types of superconductor. There is no risk of re-identifying people.

### 7. Who could benefit from your findings?

Materials scientists and chemists: ideas about which compositions to screen first, which can save lab time.

Industry in energy, MRI, electronics and transport: a faster search for practical (77 K+) superconductors.

Students and researchers: an example of careful evaluation (grouped splits, checking for leakage).


### 8. Who could be harmed or misrepresented?

Interpretation:

Researchers or funders could be misled if the results are overclaimed, e.g. calling a correlation a discovery or trusting a model that only detects cuprates. That could waste time and money.
Less-studied material families could be wrongly written off because they are underrepresented.
The original authors could be misrepresented if we don’t credit SuperCon and Hamidieh properly.

Known: No individual faces privacy loss, discrimination or stigma.

Unknown: How others might reuse our model outside its intended scope, examples may include Military weapons, surveliance technology ect.
